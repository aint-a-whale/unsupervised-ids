```python
import os

import pandas as pd
from tqdm import tqdm
```

```python
dataset_dir = 'datasets/2018/combined/'
```

```python
for dataset_filename in tqdm([i for i in os.listdir(dataset_dir) if i.endswith('.csv')]):
    dataset_path_csv = os.path.join(dataset_dir, dataset_filename)

    dataset = pd.read_csv(dataset_path_csv)

    dataset_path_parquet = os.path.join(dataset_dir, dataset_filename.replace('.csv', '.parquet'))

    dataset.to_parquet(path=dataset_path_parquet, index=False)
```
