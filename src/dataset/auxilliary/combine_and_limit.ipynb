```python
import json
import os
from pathlib import Path

import pandas as pd
from tqdm import tqdm
```

```python
pd.options.display.max_columns = 80
```

```python
def reduce_size_by_label(data: pd.DataFrame, n_rows_per_label: int) -> pd.DataFrame:
    grouped_data = data.groupby('Label', group_keys=False)
    reduced_data = grouped_data.apply(lambda x: x.sample(min(n_rows_per_label, len(x))))
    return reduced_data
```

```python
dataset_dir_path = 'datasets/2018'
combined_csv_dir_path = 'datasets/2018/combined'
combined_csv_path = os.path.join(combined_csv_dir_path, '10000.csv')
dtypes_json_path = 'datasets/2018/dtypes.json'
n_random_rows_per_label = 10000

os.makedirs(combined_csv_dir_path, exist_ok=True)
```

```python
with open(dtypes_json_path) as file:
    dtypes = json.load(file)
```

```python
combined_csv_path = 'datasets/2018/combined/10000.csv'

combined_data = pd.DataFrame()

for csv_file in tqdm(Path(dataset_dir_path).iterdir()):
    if csv_file.suffix.lower() == '.csv':
        df = pd.read_csv(csv_file, low_memory=False)
        combined_data = pd.concat([combined_data, pd.read_csv(csv_file, dtype=dtypes)], ignore_index=True)

reduced_combined_data = reduce_size_by_label(combined_data, n_random_rows_per_label)

reduced_combined_data.to_csv(combined_csv_path, index=False)

print(f'new dataset_size is {len(combined_data)}')
```
