```python
import json
import os
from pathlib import Path

import numpy as np
import pandas as pd
from tqdm import tqdm
```

```python
dataset_dir_path = 'datasets/2018'
fixed_dataset_dir_path = 'datasets/2018/fixed/'
os.makedirs(fixed_dataset_dir_path, exist_ok=True)
```

```python
total = len(os.listdir(dataset_dir_path))

for csv_file in tqdm(iterable=Path(dataset_dir_path).iterdir(), total=total):
    if csv_file.suffix.lower() == '.csv':
        df = pd.read_csv(csv_file, low_memory=False)
        df = df[df.Label != 'Label']
        df = df.replace([np.inf, -np.inf], np.nan).dropna().reset_index(drop=True)
        df = df.convert_dtypes().apply(pd.to_numeric, errors='ignore')

        flag_columns = [col for col in df.columns if 'Flag' in col]
        for col in flag_columns:
            df[col] = df[col].astype(bool)

        save_path = Path(fixed_dataset_dir_path, csv_file.name)
        df.to_csv(path_or_buf=save_path, index=False)
```

```python
for csv_file in Path(dataset_dir_path).iterdir():
    if csv_file.suffix.lower() == '.csv':
        df = pd.read_csv(csv_file, low_memory=False)
        df = df.dropna()
        df = df[df.Label != 'Label'].reset_index(drop=True)
        df = df.convert_dtypes().apply(pd.to_numeric, errors='ignore')

        save_path = Path(fixed_dataset_dir_path, 'dtypes.json')

        flag_columns = [col for col in df.columns if 'Flag' in col]
        for col in flag_columns:
            df[col] = df[col].astype(bool)

        dtypes = df.dtypes
        dtypes_dict = {col: str(dtype) for col, dtype in dtypes.items()}

        with open(save_path, 'w') as json_file:
            json.dump(dtypes_dict, json_file)
        break
```
